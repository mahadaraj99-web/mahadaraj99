## Task 1: اقرأ الجدول قبل أن تشغله

### 1. جدول التوقع قبل التشغيل

| السطر | توقعي | الناتج الفعلي | سبب أي اختلاف |
| :--- | :--- | :--- | :--- |
| `print(type(df["score"]))` | Series | `<class 'pandas.core.series.Series'>` | لا يوجد اختلاف، عمود واحد في Pandas يكون Series. |
| `print(type(df[["score"]]))` | DataFrame | `<class 'pandas.core.frame.DataFrame'>` | لا يوجد اختلاف، القوسين المزدوجين يرجّعان جدول DataFrame. |
| `print(df.loc["b", "city"])` | Jeddah | `Jeddah` | لا يوجد اختلاف، الصف `b` والعمود `city` قيمته Jeddah. |
| `print(df.iloc[1, 1])` | Jeddah | `Jeddah` | لا يوجد اختلاف، الصف الثاني (اندكس 1) والعمود الثاني (اندكس 1) هو Jeddah. |
| `print(df.loc["a":"c", "score"].tolist())` | `[88, 92, 75]` | `[88, 92, 75]` | لا يوجد اختلاف، `loc` تشمل النهاية `"c"` لأنها تعتمد على أسماء الفهرس. |
| `print(df.iloc[0:2, 2].tolist())` | `[88, 92]` | `[88, 92]` | لا يوجد اختلاف، `iloc` تستبعد النهاية `2` فتأخذ 0 و1، والعمود 2 هو `score`. |
| `print(df[df["score"] > 80]["name"].tolist())` | `['Sara', 'Noura']` | `['Sara', 'Noura']` | لا يوجد اختلاف، درجاتهم أكبر من 80. |
| `print(df["city"].value_counts()["Riyadh"])` | `2` | `2` | لا يوجد اختلاف، Riyadh تكررت مرتين. |

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["Sara", "Noura", "Reem", "Hind"],
    "city": ["Riyadh", "Jeddah", "Riyadh", "Dammam"],
    "score": [88, 92, 75, 60],
}, index=["a", "b", "c", "d"])

print(type(df["score"]))
print(type(df[["score"]]))
print(df.loc["b", "city"])
print(df.iloc[1, 1])
print(df.loc["a":"c", "score"].tolist())
print(df.iloc[0:2, 2].tolist())
print(df[df["score"] > 80]["name"].tolist())
print(df["city"].value_counts()["Riyadh"])
```

### 2. الفرق بين `df["score"]` و `df[["score"]]`
`df["score"]` يعطيني عمود واحد كـ `Series` (سلسلة بدون شكل جدول كامل)، بينما `df[["score"]]` يعطيني النتيجة كـ `DataFrame` (جدول حتى لو عمود واحد). أحتاج `df["score"]` إذا كنت أبغى أتعامل معه كعمود بسيط أو أطلع قيمة مفردة، وأحتاج `df[["score"]]` إذا كنت أبغى أحافظ على شكل جدول أو أضيفه أو أمرره لدالة تتوقع DataFrame.

### 3. ليش `df.loc["a":"c"]` رجّع 3 صفوف و `df.iloc[0:2]` رجّع صفين؟
لأن `loc` تشتغل بالأسماء (labels) وتضم بداية ونهاية المجال (`a` إلى `c` يعني a, b, c ثلاث صفوف)، أما `iloc` فتشتغل بالأرقام والـ slicing العادي في بايثون اللي يستبعد الرقم الأخير (`0:2` يعني 0 و1 فقط، صفين).

### 4. الفرق بين NumPy array و DataFrame وليه نحتاج Pandas؟
NumPy array مصفوفة أرقام سريعة جدًا، وغالبًا تكون بنفس النوع وتعتمد على أرقام الصفوف والأعمدة. أما DataFrame فهو جدول فيه صفوف وأعمدة ممكن يكون لكل عمود نوع مختلف (نصوص، أرقام، تواريخ) وله أسماء أعمدة وفهارس. نحتاج Pandas لأن بيانات الحياة الواقعية فيها أسماء أعمدة وقيم ناقصة ونصوص وتصنيفات، والتعامل معها في DataFrame أسهل وأوضح بكثير.

### 5. مثال من حياتي اليومية (قائمة مشتريات / مصروف شهري)
لو سويت جدول مشتريات:
- **الصفوف:** كل عملية شراء أو غرض اشتريته (مثلاً صف للخبز، صف للحليب، صف للبنزين).
- **الأعمدة:** معلومات هذا الغرض مثل اسم الغرض، التاريخ، السعر، الكمية، الملاحظات.
- **العمود اللي غالباً فيه قيم ناقصة:** عمود "الملاحظات" أو "الخصم"، لأني مو كل مرة أكتب ملاحظة أو يكون فيه خصم على الغرض.

## Task 3: نظف السجل وسجل قراراتك

```python
import numpy as np
import pandas as pd

data = {
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108, 105, 109, 110, 111],
    "customer": ["Sara", "Noura", "Reem", "Hind", "Lama", "Maha", "Rana", "Dana", "Lama", "Joud", "Hala", "Amal"],
    "city": ["Riyadh", "riyadh ", "Jeddah", "Dammam", "Jeddah", "RIYADH", "Riyadh", None, "Jeddah", "Dammam", "Riyadh", "Jeddah"],
    "drink": ["Latte", "Mocha", "Latte", "Espresso", "Latte", "Mocha", "Latte", "Espresso", "Latte", "Latte", None, "Mocha"],
    "price": ["18", "21", "18", "12", "18", "21", "eighteen", "12", "18", "18", "21", "21"],
    "quantity": [2, 1, 3, 1, 2, None, 1, 4, 2, 1, 2, 1],
    "rating": [5, 4, 5, 3, 4, 5, 4, None, 4, 5, 30, 4],
}

df = pd.DataFrame(data)

# 1. عدد القيم الناقصة في كل عمود وعدد الصفوف المكررة
print("القيم الناقصة قبل التنظيف:")
print(df.isna().sum())
print("\nعدد الصفوف المكررة:", df.duplicated().sum())
print("الصف المكرر:")
print(df[df.duplicated(keep=False)])

# 2. حذف الصفوف المكررة
df = df.drop_duplicates()
print("\nعدد الصفوف المكررة بعد الحذف:", df.duplicated().sum())
print("عدد الصفوف الآن:", len(df))

# 3. توحيد كتابة المدن
df["city"] = df["city"].str.strip().str.title()
print("\nقيم المدن بعد التوحيد:")
print(df["city"].value_counts(dropna=False))

# 4. تحويل عمود السعر إلى رقم (تصحيح eighteen ثم float)
df["price"] = df["price"].replace("eighteen", "18").astype(float)
print("\nنوع عمود السعر وقيمه بعد التحويل:")
print(df["price"].dtype)
print(df["price"].tolist())

# 5. التقييم من 1 إلى 5: تحويل القيمة المستحيلة (30) إلى قيمة ناقصة
df.loc[(df["rating"] < 1) | (df["rating"] > 5), "rating"] = np.nan
print("\nقيم التقييم بعد معالجة القيمة المستحيلة:")
print(df["rating"].tolist())

# 6. حساب وسيط الكمية وطباعته، ثم تعبئة القيم الناقصة
qty_median = df["quantity"].median()
print("\nوسيط الكمية (median):", qty_median)

df["quantity"] = df["quantity"].fillna(qty_median)
df["rating"] = df["rating"].fillna(df["rating"].median())
df["city"] = df["city"].fillna(df["city"].mode()[0])
df["drink"] = df["drink"].fillna("Unknown")

# 7. التأكد من عدم وجود قيم ناقصة وطباعة الجدول النظيف
print("\nالقيم الناقصة بعد التنظيف:")
print(df.isna().sum())
print("\nالجدول النظيف:")
print(df)
```

## Task 4: تقرير لصاحب الكافيه

```python
# نكمل على df النظيف من Task 3

# 1. إنشاء عمود revenue وحساب إجمالي الإيرادات
df["revenue"] = df["price"] * df["quantity"]
total_revenue = df["revenue"].sum()
print("إجمالي الإيرادات:", total_revenue)

# 2. إيرادات كل مدينة مرتبة تنازلياً
city_revenue = df.groupby("city")["revenue"].sum().sort_values(ascending=False)
print("\nإيرادات كل مدينة (تنازلي):")
print(city_revenue)

# 3. أكثر مشروب مبيعاً بالكمية، ومتوسط التقييم لكل مشروب
drink_qty = df.groupby("drink")["quantity"].sum().sort_values(ascending=False)
drink_rating = df.groupby("drink")["rating"].mean()
print("\nمبيعات كل مشروب بالكمية:")
print(drink_qty)
print("\nمتوسط التقييم لكل مشروب:")
print(drink_rating)

# 4. نسبة الطلبات من كل مدينة
city_share = df["city"].value_counts(normalize=True) * 100
print("\nنسبة الطلبات من كل مدينة (%):")
print(city_share)

# 5. أعلى طلب من حيث الإيراد: العميل والمشروب
top_order = df.loc[df["revenue"].idxmax()]
print("\nأعلى طلب من حيث الإيراد:")
print("العميل:", top_order["customer"], "| المشروب:", top_order["drink"], "| الإيراد:", top_order["revenue"])
```

- **أنصح** بعمل عروض أو باقات خاصة لفرع/طلبات **Riyadh** لأنها حققت أعلى إيراد بقيمة **117.0** وتمثل **45.45%** من إجمالي الطلبات.
- **أنصح** بمراجعة تجربة طلبات **Espresso** وتحسين جودته لأن متوسط تقييمه **3.5** وهو أقل من باقي المشروبات المقيّمة.

## Task 5: وكالة السفر عبر الزمن

```python
import numpy as np
import pandas as pd

trips = {
    "traveler": ["Zara", "Omar", "Lina", "Fahad", "Zara", "Nora", "Yousef", "Salma", "Khalid", "Maya"],
    "departure_year": [2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026, 2026],
    "destination_year": ["1450", "3026", "-300", "1999", "1450", None, "2500", "1000000", "1917", "2020"],
    "age": [29, 35, 41, 27, 29, 33, 38, 5, 45, 31],
    "price": [12000, 30000, 45000, 8000, 12000, 22000, 999999, 15000, 12000, 9000],
    "status": ["done", "done", "cancelled", "done", "done", "DONE", "done", "done", "Done", "done"],
}

df_trips = pd.DataFrame(trips)

# 1. EDA: القيم الناقصة، المكررة، nunique، value_counts للحالة
print("--- القيم الناقصة ---")
print(df_trips.isna().sum())

print("\n--- عدد الصفوف المكررة ---")
print(df_trips.duplicated().sum())
print(df_trips[df_trips.duplicated(keep=False)])

print("\n--- عدد القيم المختلفة (nunique) في كل عمود ---")
print(df_trips.nunique(dropna=False))

print("\n--- value_counts لعمود status ---")
print(df_trips["status"].value_counts(dropna=False))

# 2. Cleaning: حذف المكرر، توحيد الحالة، تحويل سنة الوجهة إلى رقم، ومعالجة الرحلة بلا وجهة والعمر السالب
df_trips = df_trips.drop_duplicates()
df_trips["status"] = df_trips["status"].str.strip().str.lower()
df_trips["destination_year"] = pd.to_numeric(df_trips["destination_year"], errors="coerce")

# الرحلة بلا وجهة (Nora) نحذفها لأنه ما نقدر نعرف سنة الوجهة أو المدة، والعمر السالب نحوله إلى NaN
df_trips = df_trips.dropna(subset=["destination_year"])
df_trips.loc[df_trips["age"] < 0, "age"] = np.nan

# 3. قرار بخصوص رحلة سنة 1,000,000 بسعر 999,999
# القرار: أعتبرها رحلة حقيقية/ممكنة في سياق وكالة سفر عبر الزمن (أو رحلة خاصة بعيدة جداً) ولا أحذفها
print("\nعدد الصفوف بعد خطوة 3:", len(df_trips))

# 4. حذف عمود departure_year لأنه ثابت (2026) في كل الصفوف ولا يضيف فرقًا في التحليل أو الموديل
df_trips = df_trips.drop(columns=["departure_year"])

# 5. إنشاء عمودين: عدد السنوات التي سافرها كل مسافر، واتجاه الرحلة (past أو future)
df_trips["years_traveled"] = (df_trips["destination_year"] - 2026).abs()
df_trips["direction"] = np.where(df_trips["destination_year"] < 2026, "past", "future")

print("\nالجدول بعد الإضافات:")
print(df_trips)

# 6. أكثر وجهة مطلوبة، متوسط السعر لكل اتجاه، عدد الرحلات المكتملة وإيرادها، وصاحب أطول رحلة
print("\n--- أكثر وجهة مطلوبة ---")
print(df_trips["destination_year"].value_counts())

print("\n--- متوسط السعر لكل اتجاه ---")
print(df_trips.groupby("direction")["price"].mean())

done_trips = df_trips[df_trips["status"] == "done"]
print("\n--- الرحلات المكتملة (done) ---")
print("عددها:", len(done_trips))
print("إجمالي إيرادها:", done_trips["price"].sum())

longest_trip = df_trips.loc[df_trips["years_traveled"].idxmax()]
print("\n--- صاحب أطول رحلة ---")
print(longest_trip)
```

### تفكير:
أصعب قرار تنظيف كان التعامل مع رحلة سنة `1000000` وسعر `999999`؛ لأنها تبدو كأنها رقم مبالغ فيه أو خطأ إدخال، لكن في سياق "وكالة سفر عبر الزمن" قد تكون رحلة بعيدة جدًا في المستقبل وسعرها مرتفع جدًا. ممكن يختلف قراري عن قرار زميلي ويكون الاثنين صحيحين لأن كل واحد يفسر البيانات حسب افتراضه: أنا اعتبرتها ضمن سيناريو القصة/البيانات المتطرفة المقبولة، بينما زميلي قد يعتبرها outlier ويستبعدها أو يحدّها.
