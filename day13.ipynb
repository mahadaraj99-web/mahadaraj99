## Task 1

**1. لماذا نرسم؟**

هذا يعني أن الأرقام الإحصائية وحدها مثل المتوسط أو الانحراف المعياري لا تكفي لفهم البيانات وقد تخدع المحلل، لأن البيانات قد تبدو متشابهة بالأرقام لكن توزيعها وشكلها الفعلي مختلف تماماً عند رسمها.

*مثال من حياتي:* إذا كان متوسط مصروفي اليومي في الأسبوع 50 ريالاً، هذا الرقم يعطي انطباعاً أنني أصرف 50 ريالاً كل يوم، لكن الحقيقة أنني لم أصرف شيئاً طوال أيام الأسبوع وصرفت 350 ريالاً دفعة واحدة في نهاية الأسبوع.

**2. اختر الرسمة:**

* **كيف تغيّرت مبيعات المقهى خلال 12 شهراً؟**
  * **الرسمة:** Line Plot (رسم خطي)
  * **لماذا؟** لأنه يوضح تغير القيم واتجاهها (صعود أو هبوط) مع مرور الوقت بشكل متصل وواضح.

* **أي فرع باع أكثر هذا الشهر من 5 فروع؟**
  * **الرسمة:** Bar Chart (رسم أعمدة)
  * **لماذا؟** لأنه الأفضل للمقارنة المباشرة بين فئات منفصلة (مثل الفروع الخمسة).

* **كيف تتوزع أعمار 500 عميل؟**
  * **الرسمة:** Histogram (مدرج تكراري)
  * **لماذا؟** لأنه يقسم الأرقام المستمرة إلى فئات ويوضح أين يتركز أغلب العملاء.

* **هل يزيد الإيجار كلما زادت مساحة الشقة؟**
  * **الرسمة:** Scatter Plot (رسم انتشار)
  * **لماذا؟** لأنه يوضح العلاقة والارتباط بين متغيرين رقميين.

* **هل في رواتب الموظفين قيم شاذة؟**
  * **الرسمة:** Box Plot (مخطط الصندوق) أو Histogram
  * **لماذا؟** لأنها تُظهر القيم البعيدة جداً عن بقية الرواتب بسهولة.

**3. أجزاء الرسمة:**

* **الفرق بين Figure و Axes:** الـ Figure هو الإطار أو اللوحة الكاملة من الخارج (والتي قد تحتوي على رسمة واحدة أو عدة رسومات)، أما الـ Axes فهي مساحة الرسم الفعلية بالداخل التي تحتوي على المحاور (X و Y) والبيانات.
* **العنوان (Title):** يوضح الفكرة العامة أو موضوع الرسمة.
* **عنوانا المحورين (xlabel / ylabel):** يوضحان ماذا يمثل كل محور وما هي وحدة القياس المستخدمة.
* **الـ Legend:** هو مفتاح الرسمة الذي يوضح معنى كل لون أو خط عندما نرسم أكثر من بيان في نفس الشكل.
* **ماذا يحدث لو أرسلتها لمديري بدونها؟** لن يفهم المدير ماذا تمثل هذه الأرقام أو عن أي موضوع تتحدث، وقد يفسر الرسمة بشكل خاطئ تماماً.

**4. رسمة مضللة:**

* **الخدعة:** المحور العمودي (Y) لم يبدأ من الصفر، بل بدأ من رقم مرتفع قريب من 94% (مثل 92%)، مما جعل الفرق البسيط (2%) يظهر بصرياً وكأنه الضعف.
* **الإصلاح:** يجب جعل محور Y يبدأ من الصفر دائماً باستخدام `plt.ylim(0, 100)` حتى تظهر الأعمدة بأحجامها الحقيقية بدون مبالغة.

## Task 2

```python
import matplotlib.pyplot as plt

days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
latte = [42, 38, 35, 40, 44, 58, 61]
espresso = [30, 28, 33, 31, 29, 35, 22]

drinks = ["Latte", "Espresso", "Mocha", "Tea"]
totals = [sum(latte), sum(espresso), 145, 96]

ratings = [5,4,4,5,3,5,4,2,5,4,4,5,3,4,5,5,1,4,4,5,
           3,4,5,4,2,5,4,4,3,5,4,5,5,4,3,4,5,4,4,4,5]

temp = [18, 22, 25, 27, 30, 33, 35, 38, 40, 41]
iced = [5, 9, 12, 15, 20, 26, 30, 35, 39, 38]
```

```python
# 1. Line Plot
plt.figure(figsize=(7, 4))
plt.plot(days, latte, marker='o', label='Latte')
plt.plot(days, espresso, marker='s', label='Espresso')
plt.title("Daily Sales: Latte vs Espresso")
plt.xlabel("Days")
plt.ylabel("Cups Sold")
plt.legend()
plt.show()
```

أعلى يوم لمبيعات الـ Latte كان يوم الجمعة (61 كوباً)، بينما انخفضت مبيعات الـ Espresso بشكل ملحوظ في نفس اليوم إلى 22 كوباً.

```python
# 2. Bar Chart
plt.figure(figsize=(7, 4))
plt.bar(drinks, totals, color=['brown', 'black', 'peru', 'green'])
plt.title("Total Weekly Sales by Drink")
plt.xlabel("Drink")
plt.ylabel("Total Cups")

for i in range(len(drinks)):
    plt.text(i, totals[i] + 3, str(totals[i]), ha='center')

plt.ylim(0, 360)
plt.show()
```

يتصدر مشروب الـ Latte المبيعات الأسبوعية بإجمالي 318 كوباً، يليه الـ Espresso بـ 208 أكواب ثم Mocha ثم Tea.

```python
# 3. Histogram
plt.figure(figsize=(7, 4))
plt.hist(ratings, bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5], edgecolor='black')
plt.title("Customer Ratings Distribution")
plt.xlabel("Rating (1 to 5)")
plt.ylabel("Number of Customers")
plt.show()
```

كما توقعت قبل التشغيل، تقييم 4 و 5 هما الأكثر تكراراً، مما يدل على رضا أغلب العملاء عن جودة المقهى.

```python
# 4. Scatter Plot
plt.figure(figsize=(7, 4))
plt.scatter(temp, iced, color='orange')
plt.title("Temperature vs Iced Coffee Sales")
plt.xlabel("Temperature (C)")
plt.ylabel("Iced Coffee Sold")
plt.show()
```

توضح الرسمة وجود علاقة طردية واضحة؛ فكلما ارتفعت درجة الحرارة زادت مبيعات القهوة المثلجة.

```python
# 5. Subplots 2x2
fig, axs = plt.subplots(2, 2, figsize=(12, 9))
fig.suptitle("Cafe Sales Dashboard", fontsize=16)

axs[0, 0].plot(days, latte, marker='o', label='Latte')
axs[0, 0].plot(days, espresso, marker='s', label='Espresso')
axs[0, 0].set_title("Daily Sales")
axs[0, 0].set_xlabel("Days")
axs[0, 0].set_ylabel("Cups")
axs[0, 0].legend()

axs[0, 1].bar(drinks, totals, color=['brown', 'black', 'peru', 'green'])
axs[0, 1].set_title("Weekly Totals")
axs[0, 1].set_xlabel("Drinks")
axs[0, 1].set_ylabel("Total Cups")

axs[1, 0].hist(ratings, bins=[0.5, 1.5, 2.5, 3.5, 4.5, 5.5], edgecolor='black')
axs[1, 0].set_title("Ratings")
axs[1, 0].set_xlabel("Stars")
axs[1, 0].set_ylabel("Count")

axs[1, 1].scatter(temp, iced, color='orange')
axs[1, 1].set_title("Temp vs Iced Coffee")
axs[1, 1].set_xlabel("Temp")
axs[1, 1].set_ylabel("Iced Cups")

plt.tight_layout()
plt.savefig("cafe_dashboard.png")
plt.show()
```

تجمع هذه اللوحة ملخص أداء المقهى بالكامل في مكان واحد، مما يسهل مقارنة المبيعات اليومية والأسبوعية وتقييمات العملاء وتأثير الطقس بسرعة.

## Task 3

```python
import pandas as pd

df = pd.read_csv("insurance.csv")
df = df.drop_duplicates()

# 1. Histogram
plt.figure(figsize=(8, 4))
plt.hist(df['charges'], bins=30, edgecolor='black', alpha=0.7)
plt.axvline(df['charges'].mean(), color='red', linestyle='--', label='Mean')
plt.axvline(df['charges'].median(), color='green', linestyle='-', label='Median')
plt.title("Distribution of Insurance Charges")
plt.xlabel("Charges")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

التوزيع ملتوي لليمين (Right-skewed)، وهذا يفسر لماذا المتوسط (الخط الأحمر) أعلى بكثير من الوسيط (الخط الأخضر)؛ لأن هناك حالات قليلة تكاليفها مرتفعة جداً وتسحب المتوسط للأعلى.

```python
# 2. Scatter
smokers = df[df['smoker'] == 'yes']
non_smokers = df[df['smoker'] == 'no']

plt.figure(figsize=(8, 5))
plt.scatter(non_smokers['age'], non_smokers['charges'], label='Non-Smoker', alpha=0.6)
plt.scatter(smokers['age'], smokers['charges'], color='red', label='Smoker', alpha=0.6)
plt.title("Age vs Charges by Smoking Status")
plt.xlabel("Age")
plt.ylabel("Charges")
plt.legend()
plt.show()
```

نرى في الرسمة 3 شرائط أو مستويات مائلة، والشريط الأعلى بالكامل يتكون من المدخنين، مما يثبت أن التدخين يرفع تكلفة التأمين بشكل كبير في جميع الأعمار.

```python
# 3. Bar Charts
fig, axs = plt.subplots(1, 2, figsize=(12, 4))

smoker_means = df.groupby('smoker')['charges'].mean()
axs[0].bar(smoker_means.index, smoker_means.values, color=['skyblue', 'salmon'])
axs[0].set_title("Average Charges by Smoker")
axs[0].set_xlabel("Smoker")
axs[0].set_ylabel("Average Charges")

region_means = df.groupby('region')['charges'].mean()
axs[1].bar(region_means.index, region_means.values, color='lightgreen')
axs[1].set_title("Average Charges by Region")
axs[1].set_xlabel("Region")
axs[1].set_ylabel("Average Charges")

plt.tight_layout()
plt.show()
```

متوسط تكلفة المدخن يقارب 4 أضعاف غير المدخن، بينما الفروقات بين المناطق الجغرافية متقاربة نسبياً.

```python
# 4. Misleading vs Accurate
fig, axs = plt.subplots(1, 2, figsize=(12, 4))

axs[0].bar(region_means.index, region_means.values, color='orange')
axs[0].set_ylim(12000, 15500)
axs[0].set_title("Misleading (Starts at 12,000)")
axs[0].set_xlabel("Region")
axs[0].set_ylabel("Charges")

axs[1].bar(region_means.index, region_means.values, color='teal')
axs[1].set_ylim(0, 16000)
axs[1].set_title("Accurate (Starts at 0)")
axs[1].set_xlabel("Region")
axs[1].set_ylabel("Charges")

plt.tight_layout()
plt.show()

ratio = region_means.max() / region_means.min()
print("النسبة الحقيقية بين أعلى منطقة وأقلها:", round(ratio, 2))
```

في الرسمة الأولى يتخيل القارئ أن منطقة southeast أعلى بـ 3 أو 4 أضعاف من southwest، بينما في الرسمة الصحيحة يتضح أن الفارق بسيط جداً والنسبة الحقيقية بينهما هي 1.19 فقط (أي زيادة 19% تقريباً).

## Task 4

```python
import numpy as np

rng = np.random.default_rng(2026)
sol = np.arange(1, 61)
temp = np.round(-60 + 15 * np.sin(sol / 9) + rng.normal(0, 4, 60), 1)
storm = np.zeros(60, dtype=int)
storm[[11, 12, 13, 32, 33, 46, 47, 48, 49]] = 1
temp[storm == 1] -= 12
wind = np.round(20 + storm * 35 + rng.normal(0, 5, 60), 1)

# 1 & 2. Temperature with Storms & Coldest Day
plt.figure(figsize=(10, 4))
plt.plot(sol, temp, label='Temperature')
plt.scatter(sol[storm == 1], temp[storm == 1], color='red', zorder=5, label='Dust Storm')

mean_temp = np.mean(temp)
plt.axhline(mean_temp, color='green', linestyle='--', label=f'Mean ({mean_temp:.1f} C)')

coldest_idx = np.argmin(temp)
coldest_sol = sol[coldest_idx]
coldest_temp = temp[coldest_idx]

plt.annotate(f'Coldest: Sol {coldest_sol} ({coldest_temp} C)',
             xy=(coldest_sol, coldest_temp),
             xytext=(coldest_sol - 12, coldest_temp - 5),
             arrowprops=dict(facecolor='black', arrowstyle='->'))

plt.title("Mars Temperature over 60 Sols")
plt.xlabel("Sol (Martian Day)")
plt.ylabel("Temperature (C)")
plt.legend()
plt.show()
```

توضح الرسمة تذبذب الحرارة على المريخ خلال 60 يوماً، ويظهر بوضوح أن أيام العواصف الترابية (النقاط الحمراء) تسبب هبوطاً حاداً في الحرارة، حيث سُجل أبرد يوم عند Sol 48.

```python
# 3. Temperature & Wind Subplots
fig, axs = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

axs[0].plot(sol, temp, color='blue')
axs[0].scatter(sol[storm == 1], temp[storm == 1], color='red')
axs[0].set_title("Mars Weather: Temperature & Wind")
axs[0].set_ylabel("Temp (C)")

colors = ['red' if s == 1 else 'gray' for s in storm]
axs[1].bar(sol, wind, color=colors)
axs[1].set_xlabel("Sol")
axs[1].set_ylabel("Wind Speed")

plt.tight_layout()
plt.show()
```

يظهر من الرسمتين تطابق تام بين هبوط درجة الحرارة في الأعلى وارتفاع سرعة الرياح (الأعمدة الحمراء) في الأسفل أثناء العواصف.

```python
# 4. Best 5 consecutive days
best_avg = -999
best_start = 0

for i in range(len(sol) - 4):
    if np.sum(storm[i:i+5]) == 0:
        avg_t = np.mean(temp[i:i+5])
        if avg_t > best_avg:
            best_avg = avg_t
            best_start = i

print(f"أفضل نافذة تبدأ من Sol {sol[best_start]} إلى Sol {sol[best_start+4]}")
print(f"متوسط الحرارة خلالها: {best_avg:.2f} C")
```

### 5. قرار الإطلاق

* **نافذة الإطلاق المختارة:** من **Sol 15 إلى Sol 19** (5 أيام متتالية).
* **متوسط حرارتها:** حوالي **-44.7 درجة مئوية**، وهي خالية تماماً من العواصف الترابية وسرعة الرياح فيها منخفضة ومستقرة.
* **لماذا ساعدتني الرسمة أسرع من الأرقام؟** النظر إلى الرسمة جعلني أستبعد فوراً الفترات التي فيها نقاط وأعمدة حمراء (العواصف) وأحدد القمة الدافئة الأولى بعيني في ثوانٍ، بدل قراءة 60 رقماً ومقارنتها يدوياً.
