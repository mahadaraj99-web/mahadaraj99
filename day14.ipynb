## Task 1

### 1. الفرق بين Matplotlib و Seaborn ولماذا نتعلم الاثنين معاً
* **Matplotlib:** هي المكتبة الأساسية للرسم في بايثون، وتعطينا تحكماً كاملاً في كل تفصيلة صغيرة في الرسمة (مثل المحاور والألوان والخطوط)، لكن كودها يكون طويلاً شوي.
* **Seaborn:** مكتبة مبنية فوق Matplotlib، تخلينا نرسم رسومات إحصائية مرتبة وشكلها حلو بسطر كود واحد فقط وتتعامل مباشرة مع جداول Pandas.
* **لماذا نتعلم الاثنين؟** نستخدم Seaborn للرسم السريع واستكشاف البيانات، ونحتاج Matplotlib عشان نعدل على الرسمة مثل إضافة عنوان (`plt.title`) أو تغيير حجم الشكل (`plt.figure`) أو عرض الرسمة (`plt.show()`).

### 2. وظيفة `hue` ومثال عليها
* **وظيفتها:** تقسم البيانات داخل نفس الرسمة إلى مجموعات وتلوّن كل مجموعة بلون مختلف بناءً على عمود فئوي، عشان نقدر نقارن بينهم بسهولة.
* **مثال:** في بيانات درجات الطلاب، أقدر أرسم العلاقة بين ساعات المذاكرة والدرجة وأحط `hue="gender"` أو `hue="study_group"`، وبكذا أكتشف هل الطلاب اللي يذاكرون في مجموعات درجاتهم أعلى من اللي يذاكرون لحالهم بنفس عدد الساعات أو لا.

### 3. الخط الأسود الصغير في `sns.barplot`
* **معناه:** يمثل **فترة الثقة (Confidence Interval - 95%)** حول المتوسط، يعني يوضح مدى تشتت البيانات أو عدم التأكد من قيمة المتوسط الحقيقي.
* **أهميته:** قبل ما نحكم إن يوم أو فئة أعلى من الثانية، لازم نشوف الخطوط السوداء؛ إذا كانت متداخلة بشكل كبير فهذا معناه إن الفرق بينهم غالباً مجرد صدفة مو فرق حقيقي ثابت.

### 4. مكونات الـ `boxplot` وعلاقتها بالـ IQR
* **الخط داخل الصندوق:** يمثل **الوسيط (Median / Q2)** وهو منتصف البيانات (50%).
* **حدود الصندوق:** الحد السفلي هو **الربيع الأول (Q1 - 25%)** والحد العلوي هو **الربيع الثالث (Q3 - 75%)**، وطول الصندوق نفسه هو المدى الربيعي `IQR = Q3 - Q1` ويحتوي على 50% من البيانات.
* **النقاط المنفصلة:** تمثل **القيم الشاذة (Outliers)** اللي تخرج برا الحدود الطبيعية (أقل من `Q1 - 1.5 * IQR` أو أكبر من `Q3 + 1.5 * IQR`).

## Task 2

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

tips = sns.load_dataset("tips")
tips.head()
```

**1. توقع قبل التشغيل:** أتوقع إن فواتير العشاء (Dinner) بتكون أكبر من الغداء (Lunch) لأن الناس غالباً يطلبون وجبات أكثر في العشاء.

```python
plt.figure(figsize=(7, 4))
sns.histplot(data=tips, x="total_bill", hue="time")
plt.title("Distribution of Total Bill by Meal Time")
plt.show()

print(tips.groupby("time", observed=False)["total_bill"].mean())
```

**النتيجة:** فعلاً فواتير العشاء (Dinner) أكبر وأكثر انتشاراً، ومتوسط الفاتورة في العشاء حوالي `20.80` دولار مقارنة بـ `17.17` دولار في الغداء.

**2. توقع قبل التشغيل:** أتوقع يوم السبت أو الأحد يكون فيه أعلى بقشيش لأنها أيام نهاية الأسبوع ويكثر فيها الزبائن.

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=tips, x="day", y="tip")
plt.title("Tip Distribution by Day")
plt.show()

print(tips.groupby("day", observed=False)["tip"].median())
```

**النتيجة:** يوم الأحد (Sun) فيه أعلى وسيط للبقشيش بقيمة `3.15` دولار، بينما يوم السبت (Sat) فيه أكثر قيم شاذة (نقاط منفصلة فوق الصندوق تصل إلى 10 دولار).

**3. توقع قبل التشغيل:** أتوقع عدد الفواتير في الويكند (السبت والأحد) يكون أكثر من باقي الأيام.

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=tips, x="day", hue="smoker")
plt.title("Number of Bills per Day by Smoker Status")
plt.show()
```

**النتيجة:** يوم السبت هو الأكثر في عدد الفواتير (87 فاتورة إجمالاً: 45 لغير المدخنين و42 للمدخنين)، ويوم الجمعة هو الأقل (19 فاتورة فقط).

**4. توقع قبل التشغيل:** أتوقع كل ما زادت قيمة الفاتورة زاد البقشيش، وكل ما كبر حجم الطاولة زادت الفاتورة.

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time", size="size")
plt.title("Total Bill vs Tip by Time and Party Size")
plt.show()

corr_val = tips["total_bill"].corr(tips["tip"])
print("Correlation:", round(corr_val, 2))
```

**النتيجة:** نعم، توجد علاقة طردية واضحة؛ كلما زادت الفاتورة زاد البقشيش، ومعامل الارتباط بينهما يساوي `0.68`.

**5. توقع قبل التشغيل:** أتوقع متوسط البقشيش متقارب بين الأيام والخطوط السوداء بتكون متداخلة.

```python
plt.figure(figsize=(7, 4))
sns.barplot(data=tips, x="day", y="tip")
plt.title("Average Tip by Day")
plt.show()
```

**النتيجة:** متوسط البقشيش يتراوح بين `2.73` و`3.26` دولار، والخطوط السوداء (فترة الثقة) متداخلة بين الأيام، وهذا يدل إن الفرق بين الأيام مو كبير جداً أو حاسم.

**6. توقع قبل التشغيل:** أتوقع أعلى ارتباط بيكون بين الفاتورة والبقشيش، وأقل ارتباط بين حجم الطاولة والبقشيش.

```python
plt.figure(figsize=(6, 4))
corr_tips = tips[["total_bill", "tip", "size"]].corr()
sns.heatmap(corr_tips, annot=True, cmap="Blues")
plt.title("Correlation Heatmap (Tips Dataset)")
plt.show()
```

**النتيجة:** أعلى ارتباط هو بين `total_bill` و`tip` بقيمة `0.68`، يليه `total_bill` و`size` بقيمة `0.60`، وأقل ارتباط بين `tip` و`size` بقيمة `0.49`.

```python
df_loan = pd.read_csv("loan_clean.csv")
df_loan.head()
```

```python
# 1. countplot للـ Target
plt.figure(figsize=(6, 4))
sns.countplot(data=df_loan, x="Loan_Status")
plt.title("Loan Approval Status Count")
plt.show()

print(df_loan["Loan_Status"].value_counts())
```

**1. عدد الطلبات:** يوضح الرسم عدد الطلبات المقبولة (Y) والمرفوضة (N)، ونلاحظ أن الطلبات المقبولة تمثل الأغلبية (حوالي 69% مقبول مقابل 31% مرفوض).

**2. نسبة الموافقة:** أصحاب السجل الائتماني الجيد (`Credit_History = 1`) نسبة قبولهم عالية جداً تتجاوز `78%` في كل المناطق، وخاصة منطقة `Semiurban` اللي تتصدر القبول، بينما اللي سجلهم `0` نسبة قبولهم منخفضة جداً (أقل من `10%`).

```python
# 4. histplot للدخل كامل وللدخل أقل من 20,000
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(data=df_loan, x="ApplicantIncome", ax=axes[0])
axes[0].set_title("All Applicant Income")

sns.histplot(data=df_loan[df_loan["ApplicantIncome"] < 20000], x="ApplicantIncome", ax=axes[1])
axes[1].set_title("Applicant Income (< 20,000)")

plt.tight_layout()
plt.show()
```

## Task 4

```python
rng = np.random.default_rng(7)
n = 150
species = rng.choice(["Fire", "Ice", "Storm"], n, p=[0.4, 0.35, 0.25])
base_wing = {"Fire": 14, "Ice": 11, "Storm": 18}
base_heat = {"Fire": 900, "Ice": 80, "Storm": 300}
dragons = pd.DataFrame({
    "dragon_id": [f"D{i:03d}" for i in range(n)],
    "species": species,
    "wingspan_m": [round(base_wing[s] + rng.normal(0, 2), 1) for s in species],
    "fire_temp_c": [round(base_heat[s] + rng.normal(0, 60)) for s in species],
    "age_years": rng.integers(20, 400, n),
})
dragons["weight_t"] = (dragons["wingspan_m"] * 0.6 + rng.normal(0, 0.8, n)).round(1)
dragons.loc[37, "wingspan_m"] = 140.0

dragons.head()
```

```python
# 1. countplot: كم تنينا من كل نوع؟
plt.figure(figsize=(6, 4))
sns.countplot(data=dragons, x="species")
plt.title("Count of Dragons by Species")
plt.show()

print(dragons["species"].value_counts())
```

**1. عدد التنانين:** يوجد `56` تنين نار (Fire)، و`54` تنين جليد (Ice)، و`40` تنين عواصف (Storm).

```python
# 2. boxplot لطول الجناح حسب النوع لاكتشاف الخطأ
plt.figure(figsize=(6, 4))
sns.boxplot(data=dragons, x="species", y="wingspan_m")
plt.title("Wingspan by Species (Before Cleaning)")
plt.show()

print(dragons[dragons["wingspan_m"] > 50])
```

**2. التنين الغريب وتصحيحه:**
* التنين الغريب هو **D037** من نوع `Fire` وطول جناحه مسجل `140.0` متر!
* **سبب الخطأ:** خطأ إدخال (فاصلة عشرية في المكان الخطأ)، لأن متوسط جناح تنانين النار حول `14.0` متر، وحتى وزنه `10.0` طن يناسب جناح طوله `14.0` متر (لأن `14 * 0.6 = 8.4` طن تقريباً).
* **التصحيح:** نعدّل القيمة إلى `14.0`.

```python
dragons.loc[37, "wingspan_m"] = 14.0

# 3. scatterplot لطول الجناح مقابل حرارة النار بعد التصحيح
plt.figure(figsize=(7, 4))
sns.scatterplot(data=dragons, x="wingspan_m", y="fire_temp_c", hue="species")
plt.title("Wingspan vs Fire Temperature by Species (Cleaned)")
plt.show()
```

**3. العلاقة بين طول الجناح وحرارة النار:** تنقسم التنانين إلى 3 مجموعات واضحة حسب النوع؛ تنانين `Fire` هي الأعلى حرارة (حول 900 درجة) ومتوسطة الجناح، وتنانين `Storm` هي الأطول جناحاً (حول 18 متر) بحرارة متوسطة (حول 300 درجة)، وتنانين `Ice` هي الأقل في الاثنين.

```python
# 4. heatmap للارتباط
plt.figure(figsize=(6, 4))
corr_dragons = dragons[["wingspan_m", "fire_temp_c", "age_years", "weight_t"]].corr()
sns.heatmap(corr_dragons, annot=True, fmt=".2f", cmap="YlOrRd")
plt.title("Correlation Heatmap (Dragons Dataset)")
plt.show()
```

**4. الارتباط بين الأعمدة:** العمودان اللذان يكادان يقيسان الشيء نفسه هما **`wingspan_m` (طول الجناح) و `weight_t` (الوزن)** بمعامل ارتباط قوي جداً يساوي **`0.93`**.
